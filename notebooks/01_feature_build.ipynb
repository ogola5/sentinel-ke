# 01 - Feature Snapshot Build

Build Layer-3 graph feature snapshots from ledger data before GNN training.

```python
import json
import sys
from sqlalchemy import text

if '/app' not in sys.path:
    sys.path.append('/app')

from app.ledger.db import SessionLocal
from app.analytics.layer3.graph_feature_worker import run_once as run_feature_snapshot
```

```python
db = SessionLocal()
try:
    rows = run_feature_snapshot(db=db, window_key='Wmid', max_entities=5000)
    print(json.dumps({'features_upserted': rows}, indent=2))
finally:
    db.close()
```

```python
db = SessionLocal()
try:
    q = text('''
        SELECT window_key, window_end, COUNT(*) AS entities
        FROM graph_feature_snapshot
        GROUP BY window_key, window_end
        ORDER BY window_end DESC
        LIMIT 10
    ''')
    rows = db.execute(q).fetchall()
    for r in rows:
        print(r)
finally:
    db.close()
```
